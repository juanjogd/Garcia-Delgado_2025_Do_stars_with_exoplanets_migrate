##### Kolmogorov-Smirnov test

First, we create the cumulative distributions for the guiding and birth radius:

```python
x = np.random.normal(0, 1, 1000)
z = np.random.normal(0, 0.9, 1000)

sns.ecdfplot(pd.DataFrame({'Birth radius': final_catalogue['Rbirth_50%'], 'Guiding radius': final_catalogue['rguiding_50%']}))
```

Now we can calculate the KS test values

```python
from scipy import stats

kstest_test = stats.kstest(final_catalogue['Rbirth_50%'], final_catalogue['rguiding_50%'])
print(kstest_test,len(final_catalogue['Rbirth_50%']),len(final_catalogue['rguiding_50%']))
```

We can also estimate ourselves the likelyhood of a null hypothesis (from wikipedia: https://en.wikipedia.org/wiki/Kolmogorov–Smirnov_test)

```python
#D> c(alpha) * sqrt((n+m)/(n*m)) for a level of alpha confidence.
# Since we have the value of D from the KS test:
c_alpha = kstest_test[0]/np.sqrt(2/1506)
print(c_alpha)
```

```python
#alpha can be defined as c(alpha)=sqrt(-ln(alpha/2)*0.5)
#therefore alpha can be calculated as: 
alpha=2.*np.exp(-2.*(11.022987**2.))
print(alpha)
```

Our value differes 3 orders of magnitude from the estimated p_value from the scipy routine, but is consistent with null at a very high probability
